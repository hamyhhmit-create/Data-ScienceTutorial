# Nested function
Hàm lồng nhau

```python
def square():
    def add():
        x = 2
        y = 3
        z = x + y
        return z
    return add()**2
print(square())
```

# Default and flexible arguments
Cách truyền đối số vào hàm

```python
# Default arguments (đối số mặc định)
def f(a, b = 1, c = 2) :
  y = a + b + c
  return y
print (f(5))
print (f(5,4,3))
```

```python
# Flexible arguments *args (đối số linh hoạt)
def f(*args) :
  for i in args:
    print(i)
f(1)
print("")
f(1,2,3,4)
```

```python
# Flexible arguments **kwargs (đối số linh hoạt)
def f(**kwargs) :
  for key, value in kwargs.items():
    print(key, " ", value)
f(country = 'spain', capital = 'madrid', population = 123456)
```

# Lambda function

```python
square = lambda x: x**2
print(square(4))
tot = lambda x,y,z: x+y+z
print(tot(1,2,3))
```

# Anonymous function
Hàm ẩn danh

```python
number_list = [1,2,3]
y = map(lambda x:x**2,number_list)
print(list(y))
```

```python
list1 = [1,2,3,4]
list2 = [5,6,7,8]
z = zip(list1,list2)
print(z)
z_list = list(z)
print(z_list)
```

```python
un_zip = zip(*z_list)
un_list1,un_list2 = list(un_zip)
print(un_list1)
print(un_list2)
print(type(un_list2))
```

# List comprehension


```python
num1 = [1,2,3]
num2 = [i + 1 for i in num1 ]
print(num2)
```

```python
num1 = [5,10,15]
num2 = [i**2 if i == 10 else i-5 if i < 7 else i+5 for i in num1]
print(num2)
```
