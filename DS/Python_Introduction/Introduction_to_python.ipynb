# Lập trình với python

```python
# khai báo thư viện
import pandas as pd # xử lý và phân tích dữ liệu dạng bảng
df = pd.read_CSV("student.csv") # đọc dữ liệu
print(df.head()) # xem dữ liệu
print(df.info()) # thông tin dữ liệu

print(df.describe()) # thống kê
high_score = df[df["Score"] > 8] # lọc sinh viên có điểm lớn hơn 8
high_score = high_score.sort_values("Score", ascending=False) # sắp xếp theo điểm giảm dần
high_score.to_csv("high_score.csv", index=False) # lưu kết quả


import numpy as np # đại số tuyến tính, mảng, thống kê
a = np.array([1,2,3,4,5]) # tạo mảng

print("Tổng: ", np.sum(a)) # thống kê
print("Trung bình: ", np.mean(a)) # thống kê
print("Lớn nhất: ", np.max(a)) # thống kê

print(a * 2) # phép toán

b = np.arange(12).reshape(3,4)
print(b) # đổi hình dạng

A = np.array([[1,2],
              [3,4]])
B = np.array([[5,6],
              [7,8]])
print(A@B) # nhân ma trận

import matplotlib.pyplot as plt # cung cấp hàm để tạo và hiển thị biểu đồ
# Biểu đồ line
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y)
plt.show()

# Biểu đồ cột
name = ["A", "B", "C"]
score = [8,6,9]
plt.bar(name,score)
plt.show()

```

# Matplotlib

```python
import matplotlib.pyplot as plt # cung cấp hàm để tạo và hiển thị biểu đồ
# Biểu đồ line
x = [1,2,3,4]
y = [2,4,6,8]
plt.plot(x,y)
plt.show()

# Biểu đồ cột
name = ["A", "B", "C"]
score = [8,6,9]
plt.bar(name,score)
plt.show()

# Biểu đồ phân tán
x = [1,2,3,4]
y = [5,7,6,8]
plt.scatter(x,y)
plt.show()

# Biểu đồ tròn
labels = ["A", "B", "C"]
sizes = [40,35,25]
plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.show()
```

# Seaborn

```python
import seaborn as sns
# Vẽ biểu đồ phân tán
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x = "total_bill", y = "tips")
plt.show()

# Vẽ biểu đồ đường
sns.lineplot(data=tips, x ="size", y = "total_bill")
plt.show()

# Vẽ biểu đồ cột
sns.barplot(data=tips, x ="day", y = "total_bill")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x="total_bill", y="tip")

plt.show()
```

```python
sns.lineplot(data=tips, x="size", y="total_bill")

plt.show()
```

```python
sns.barplot(data=tips, x = "day", y = "total_bill")
plt.show()
```

```python
sns.boxplot(data=tips, x="day", y="total_bill")

plt.show()
```
