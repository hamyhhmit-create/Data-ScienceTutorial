```python
import gdown
import pandas as pd  # phân tích dữ liệu dạng bảng
import matplotlib.pyplot as plt  # vẽ biểu đồ

# 1. Tải file từ Google Drive về máy/Colab bằng gdown
url = 'https://drive.google.com/uc?id=17SBKRXRljjEWWsxSDVYQ4LIgriYsIeyI'
gdown.download(url, 'pokemon.csv', quiet=False)

# 2. Đọc file CSV đã tải về
data = pd.read_csv('pokemon.csv')

# Line Plot
# color = color, label = label, linewidth = width of line, alpha = opacity, grid = grid, linestyle = sytle of line
data.Speed.plot(kind = 'line', color = 'g',label = 'Speed',linewidth=1,alpha = 0.5,grid = True,linestyle = ':')
data.Defense.plot(color = 'r',label = 'Defense',linewidth=1, alpha = 0.5,grid = True,linestyle = '-.')
plt.legend(loc='upper right')     # legend = puts label into plot
plt.xlabel('x axis')              # label = name of label
plt.ylabel('y axis')
plt.title('Line Plot')            # title = title of plot

plt.show() # hiển thị biểu đồ
```

```python
# Scatter Plot
# x = attack, y = defense
data.plot(kind='scatter', x='Attack', y='Defense',alpha = 0.5,color = 'red')
plt.xlabel('Attack')              # label = name of label
plt.ylabel('Defence')
plt.title('Attack Defense Scatter Plot')            # title = title of plot
```

```python
# Histogram
# bins = number of bar in figure
data.Speed.plot(kind = 'hist',bins = 50,figsize = (12,12))
plt.show()
```

```python
# clf() = cleans it up again you can start a fresh
data.Speed.plot(kind = 'hist',bins = 50)
plt.clf()
# We cannot see plot due to clf()
```

# Dictionary


```python
#create dictionary and look its keys and values
dictionary = {'spain' : 'madrid','usa' : 'vegas'} # keys: values
print(dictionary.keys())
print(dictionary.values())
```

```python
# Immutable (đối tượng bất biến): string, boolean, float, integer or tubles
# List không thể thay
# Mỗi keys là duy
dictionary['spain'] = "barcelona"    # Cập nhật một phần tử đã tồn tại
print(dictionary)
dictionary['france'] = "paris"       # Cập nhật một phần tử
print(dictionary)
del dictionary['spain']              # remove entry with key 'spain'
print(dictionary)
print('france' in dictionary)        # Check xem france có nằm trong dictionary không
dictionary.clear()                   # Xóa toàn bộ
print(dictionary)
```

# Seaborn

```python
import seaborn as sns
# Vẽ biểu đồ phân tán
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x = "total_bill", y = "tips")
plt.show()

# Vẽ biểu đồ đường
sns.lineplot(data=tips, x ="size", y = "total_bill")
plt.show()

# Vẽ biểu đồ cột
sns.barplot(data=tips, x ="day", y = "total_bill")
plt.show()
```
