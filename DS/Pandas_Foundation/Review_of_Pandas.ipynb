# Review of Pandas

# Building data frames from scratch

```python
import pandas as pd
country = ["Spain","France"]
population = ["11","12"]
list_label = ["country","population"]
list_col = [country,population]
zipped = list(zip(list_label,list_col))
data_dict = dict(zipped)
df = pd.DataFrame(data_dict)
df
```

```python
df["capital"] = ["madrid","paris"]
df
```

```python
df["income"] = 0
df
```

```python
import gdown
import pandas as pd
import matplotlib.pyplot as plt

url = 'https://drive.google.com/uc?id=17SBKRXRljjEWWsxSDVYQ4LIgriYsIeyI'
gdown.download(url, 'pokemon.csv', quiet=False)

data = pd.read_csv('pokemon.csv')

data1 = data.loc[:,["Attack","Defense","Speed"]]
data1.plot()
```

```python
data1.plot(subplots = True)
plt.show()
```

```python
data1.plot(kind = "scatter",x="Attack",y = "Defense")
plt.show()
```

```python
data1.plot(
    kind="hist",
    y="Defense",
    bins=50,
    range=(0,250),
    density=True
)
```

```python
fig, axes = plt.subplots(nrows=2, ncols=1)

data1.plot(
    kind="hist",
    y="Defense",
    bins=50,
    range=(0, 250),
    density=True,
    ax=axes[0]
)

data1.plot(
    kind="hist",
    y="Defense",
    bins=50,
    range=(0, 250),
    density=True,
    ax=axes[1],
    cumulative=True
)

plt.savefig("graph.png")
plt.show()
```

# Indexing pandas time series

```python
time_list = ["1992-03-08","1992-04-12"]
print(type(time_list[1]))
datetime_object = pd.to_datetime(time_list)
print(type(datetime_object))
```

```python
import warnings
warnings.filterwarnings("ignore")
data2 = data.head()
date_list = ["1992-01-10","1992-02-10","1992-03-10","1993-03-15","1993-03-16"]
datetime_object = pd.to_datetime(date_list)
data2["date"] = datetime_object
data2= data2.set_index("date")
data2
```

```python
print(data2.loc["1993-03-16"])
print(data2.loc["1992-03-10":"1993-03-16"])
```

# Resampling pandas time series

```python
data2.resample("YE").mean(numeric_only=True)
```

```python
data2.resample("ME").mean(numeric_only=True)
```

```python
data2.resample("ME").first().interpolate(method="linear")
```

```python
data2.resample("ME").mean(numeric_only=True).interpolate(method="linear")
```
